```python
print("Họ tên: Lâm Đặng Gia Kiện - MSSV: 6351071036")
print("Câu 3 - Bài 1 - Phần 2")
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# Bài 3: CSDL Nhân viên (Tập trung vào Tuổi, Năng lực, Thăng chức)
data_3 = {
    'Tuổi': ['20..25', '20..25', '26..30', '31..40', '26..30', '26..30', '31..40', '26..30', '>40', '26..30'],
    'Năng lực': ['Giỏi', 'Khá', 'Giỏi', 'T.Bình', 'T.Bình', 'Khá', 'Khá', 'Khá', 'Giỏi', 'Giỏi'],
    'Thăng chức': ['Có', 'Không', 'Có', 'Có', 'Không', 'Không', 'Không', 'Có', 'Không', 'Có']
}
df3 = pd.DataFrame(data_3)

# One-hot encoding các giá trị
df3_encoded = pd.get_dummies(df3).astype(bool)

print("--- Tập phổ biến (min_sup=30%) ---")
freq_items_3 = apriori(df3_encoded, min_support=0.30, use_colnames=True)
display(freq_items_3)

print("\n--- Luật kết hợp dẫn đến Thăng chức (min_sup=30%, min_conf=80%) ---")
try:
    # Tìm tập luật
    rules_3 = association_rules(freq_items_3, metric="confidence", min_threshold=0.80)

    # Kiểm tra xem có luật nào được tìm thấy không
    if rules_3.empty:
        print("=> KHÔNG TÌM THẤY LUẬT NÀO thỏa mãn cả ngưỡng min_sup=30% và min_conf=80%.")
    else:
        # Lọc các luật CHỈ có 'Thăng chức' ở vế phải (consequents)
        def is_promotion(consequents):
            return any('Thăng chức' in item for item in consequents)

        filtered_rules = rules_3[rules_3['consequents'].apply(is_promotion)]

        if filtered_rules.empty:
            print("=> Không có luật nào có vế phải là 'Thăng chức' thỏa mãn điều kiện.")
        else:
            display(filtered_rules[['antecedents', 'consequents', 'support', 'confidence']])

except ValueError:
    print("=> KHÔNG TÌM THẤY LUẬT NÀO thỏa mãn cả ngưỡng min_sup=30% và min_conf=80%.")
```
