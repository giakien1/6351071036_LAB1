```python
# CHẠY Ô NÀY ĐỂ CHUẨN BỊ DỮ LIỆU CHO BÀI 5
print("Họ tên: Lâm Đặng Gia Kiện - MSSV: 6351071036")
try:
    df_retail = pd.read_excel('Online Retail.xlsx')
    df_retail['Description'] = df_retail['Description'].str.strip()
    df_retail.dropna(axis=0, subset=['InvoiceNo'], inplace=True)
    df_retail['InvoiceNo'] = df_retail['InvoiceNo'].astype('str')
    df_retail = df_retail[~df_retail['InvoiceNo'].str.contains('C')]
    print("Đã load và làm sạch dữ liệu Online Retail thành công!")
except FileNotFoundError:
    print("LỖI: Vui lòng tải file 'Online Retail.xlsx' lên Colab!")
```
