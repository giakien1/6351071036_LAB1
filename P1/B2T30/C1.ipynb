```python
print("Họ tên: Lâm Đặng Gia Kiện - MSSV: 6351071036")
print("Câu 1 - Bài 2 Trang 30 - Phần 1")
import pandas as pd
import matplotlib.pyplot as plt

# Đọc dữ liệu từ tập tin đính kèm (không dùng dòng đầu làm header)
df_track = pd.read_excel('tb_tracking.xlsx', header=None)

print("========== a) THÊM TIÊU ĐỀ CỘT ==========")
# Thêm vào tiêu đề các cột theo thứ tự yêu cầu
df_track.columns = ['id', 'device_id', 'co_level', 'humidity', 'temperature', 'time']
display(df_track.head())


print("\n========== b) LỌC DỮ LIỆU TỪ NGÀY 03/01/2018 ==========")
# Chuyển đổi cột 'time' sang định dạng datetime của pandas
df_track['time'] = pd.to_datetime(df_track['time'])

# Lọc bỏ những dữ liệu trước ngày 03/01/2018
df_track = df_track[df_track['time'] >= '2018-01-03']
print(f"Số dòng dữ liệu sau khi lọc: {len(df_track)}")
display(df_track.head())


print("\n========== c) BÓC TÁCH THỜI ĐIỂM ĐO ĐẠC THÀNH PHÚT ==========")
# Tạo cột 'minutes' bằng cách lấy (Giờ * 60) + Phút
df_track['minutes'] = df_track['time'].dt.hour * 60 + df_track['time'].dt.minute
display(df_track[['time', 'minutes']].head())


print("\n========== d) VẼ BIỂU ĐỒ PHÂN TÁN (SCATTER-PLOTS) ==========")
# Thiết lập không gian vẽ biểu đồ (1 hàng, 3 cột)
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Biểu đồ Nồng độ khí CO
axes[0].scatter(df_track['minutes'], df_track['co_level'], alpha=0.5, color='blue')
axes[0].set_title('Nồng độ khí CO theo phút')
axes[0].set_xlabel('Thời gian (phút trong ngày)')
axes[0].set_ylabel('Nồng độ CO')

# Biểu đồ Độ ẩm
axes[1].scatter(df_track['minutes'], df_track['humidity'], alpha=0.5, color='green')
axes[1].set_title('Độ ẩm theo phút')
axes[1].set_xlabel('Thời gian (phút trong ngày)')
axes[1].set_ylabel('Độ ẩm (%)')

# Biểu đồ Nhiệt độ
axes[2].scatter(df_track['minutes'], df_track['temperature'], alpha=0.5, color='red')
axes[2].set_title('Nhiệt độ theo phút')
axes[2].set_xlabel('Thời gian (phút trong ngày)')
axes[2].set_ylabel('Nhiệt độ (°C)')

plt.tight_layout()
plt.show()
```
